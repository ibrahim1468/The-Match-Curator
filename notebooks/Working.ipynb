```python
def get_k_value(tournament_name):
    tournament = str(tournament_name).lower().strip()

    if "fifa world cup" in tournament and "qualification" not in tournament:
        return 50
    elif "qualification" in tournament:
        return 30
    elif "copa américa" in tournament or "copa america" in tournament:
        return 35
    elif "african cup of nations" in tournament:
        return 35
    elif "uefa euro" in tournament:
        return 35
    elif "afc asian cup" in tournament:
        return 35
    elif "gold cup" in tournament:
        return 35
    elif "uefa nations league" in tournament:
        return 30
    elif "concacaf nations league" in tournament:
        return 30
    elif "friendly" in tournament:
        return 10
    else:
        return 20

print(get_k_value("FIFA World Cup"))
print(get_k_value("FIFA World Cup qualification"))
print(get_k_value("Friendly"))
print(get_k_value("UEFA Euro"))
print(get_k_value("CECAFA Cup"))
print(get_k_value("Copa América"))
```

```python
elo_dict = {}
unique_teams = pd.unique(df_elo[["home_team", "away_team"]].values.ravel())
for i in unique_teams:
    elo_dict[i] = 1000

print(len(elo_dict))
print(elo_dict["France"])
print(elo_dict["Brazil"])
```

```python
for _, row in df_elo.iterrows():

    # Match information
    home_team = row["home_team"]
    away_team = row["away_team"]
    home_score = row["home_score"]
    away_score = row["away_score"]
    tournament = row["tournament"]

    # Current ELO ratings
    elo_home = elo_dict[home_team]
    elo_away = elo_dict[away_team]

    # Expected scores
    expected_home = 1 / (1 + 10 ** ((elo_away - elo_home) / 400))
    expected_away = 1 - expected_home

    # Actual scores
    if home_score > away_score:
        actual_home = 1.0
    elif home_score == away_score:
        actual_home = 0.5
    else:
        actual_home = 0.0

    actual_away = 1 - actual_home

    # K factor
    K = get_k_value(tournament)

    # Update ratings
    elo_dict[home_team] = elo_home + K * (actual_home - expected_home)
    elo_dict[away_team] = elo_away + K * (actual_away - expected_away)
```

```python
global_chaos = (df_wc["home_score"] + df_wc["away_score"]).mean()
print(round(global_chaos, 2))
```

```python
chaos_dict = {}

for team in wc2026_teams:

    team_matches = df_wc[
        (df_wc["home_team"] == team) |
        (df_wc["away_team"] == team)
    ]

    total_matches = len(team_matches)

    if total_matches == 0:
        # edge case: team never appears (shouldn't happen, but datasets love surprises)
        chaos_score = global_chaos
    else:
        chaos_score = (team_matches["home_score"] + team_matches["away_score"]).mean()

    # shrinkage rule
    if total_matches < 10:
        chaos_score = global_chaos

    chaos_dict[team] = chaos_score

print(len(chaos_dict))

sorted_chaos = sorted(chaos_dict.items(), key=lambda x: x[1], reverse=True)
for team, score in sorted_chaos[:5]:
    print(f"{team}: {round(score, 2)}")

print("---")
for team, score in sorted_chaos[-5:]:
    print(f"{team}: {round(score, 2)}")
```

```python
chaos_values = list(chaos_dict.values())

min_chaos = min(chaos_values)
max_chaos = max(chaos_values)

chaos_normalized = {}

for team, value in chaos_dict.items():
    normalized = (value - min_chaos) / (max_chaos - min_chaos)
    chaos_normalized[team] = normalized

print(min(chaos_normalized.values()))
print(max(chaos_normalized.values()))
```

```python
df_form = df_result.dropna(subset=["home_score", "away_score"])
```

```python
form_dict = {}

for team in wc2026_teams:

    team_matches = df_form[
        (df_form["home_team"] == team) |
        (df_form["away_team"] == team)
    ]

    team_matches = team_matches.sort_values("date").tail(10)

    results = []

    for _, row in team_matches.iterrows():

        home = row["home_team"]
        away = row["away_team"]
        hs = row["home_score"]
        as_ = row["away_score"]

        if home == team:
            if hs > as_:
                results.append(1.0)
            elif hs == as_:
                results.append(0.5)
            else:
                results.append(0.0)

        else:  # team is away
            if as_ > hs:
                results.append(1.0)
            elif as_ == hs:
                results.append(0.5)
            else:
                results.append(0.0)

    # average form (handle edge case)
    if len(results) == 0:
        form_score = 0.5  # neutral fallback
    else:
        form_score = sum(results) / len(results)

    form_dict[team] = form_score
```

```python
sorted_form = sorted(form_dict.items(), key=lambda x: x[1], reverse=True)
for team, score in sorted_form[:5]:
    print(f"{team}: {round(score, 2)}")
print("---")
for team, score in sorted_form[-5:]:
    print(f"{team}: {round(score, 2)}")
```

```python
form_values = list(form_dict.values())

min_form = min(form_values)
max_form = max(form_values)

form_normalized = {}

for team, value in form_dict.items():
    if max_form == min_form:
        form_normalized[team] = 0.0
    else:
        form_normalized[team] = (value - min_form) / (max_form - min_form)

print(min(form_normalized.values()))
print(max(form_normalized.values()))
print(round(form_normalized["France"], 3))
print(round(form_normalized["New Zealand"], 3))
```

```python
rows = []

importance_map = {
    "group stage": 0.4,
    "Round of 32": 0.6,
    "Round of 16": 0.75,
    "Quarter-final": 0.85,
    "Semi-final": 0.9,
    "Play-off for third place": 0.6,
    "Final": 1.0
}

for _, row in df_schedule.iterrows():

    team1 = row["team1"]
    team2 = row["team2"]

    if team1 == "TBD" or team2 == "TBD":
        continue

    elo_gap_raw = abs(elo_wc_teams[team1] - elo_wc_teams[team2])

    chaos_combined = (chaos_normalized[team1] + chaos_normalized[team2]) / 2
    form_combined = (form_normalized[team1] + form_normalized[team2]) / 2
    star_combined = (star_normalized[team1] + star_normalized[team2]) / 2

    rivalry_score_raw = rivalries.get(frozenset({team1, team2}), 0)

    match_importance = importance_map.get(row["stage"], 0.4)

    rows.append({
        "match_id": row["match_id"],
        "date": row["date"],
        "team1": team1,
        "team2": team2,
        "group": row.get("group", None),
        "stage": row["stage"],

        "elo_gap": elo_gap_raw,
        "rivalry_score": rivalry_score_raw,

        "chaos_combined": chaos_combined,
        "form_combined": form_combined,
        "star_combined": star_combined,

        "match_importance": match_importance
    })

df_features = pd.DataFrame(rows)
```

```python
# ---- ELO GAP NORMALIZATION ----
min_gap = df_features["elo_gap"].min()
max_gap = df_features["elo_gap"].max()

df_features["elo_gap_norm"] = df_features["elo_gap"].apply(
    lambda x: 0.0 if max_gap == min_gap else (x - min_gap) / (max_gap - min_gap)
)

df_features["elo_closeness"] = 1 - df_features["elo_gap_norm"]


# ---- RIVALRY NORMALIZATION (NEW FIX) ----
min_r = df_features["rivalry_score"].min()
max_r = df_features["rivalry_score"].max()

df_features["rivalry_norm"] = df_features["rivalry_score"].apply(
    lambda x: 0.0 if max_r == min_r else (x - min_r) / (max_r - min_r)
)

max_rivalry = df_features["rivalry_score"].max()
if max_rivalry > 0:
    df_features["rivalry_score"] = df_features["rivalry_score"] / max_rivalry
else:
    df_features["rivalry_score"] = 0.0
```

```python
hype_bonus = {
    frozenset({"Norway", "France"}): 0.06,
    frozenset({"Brazil", "Morocco"}): 0.04,
    frozenset({"Argentina", "Algeria"}): 0.05,
    frozenset({"England", "Croatia"}): 0.04,
    frozenset({"Germany", "Portugal"}): 0.05,
    frozenset({"Spain", "France"}): 0.05,
}

def get_hype(team1, team2):
    return hype_bonus.get(frozenset({team1, team2}), 0)

df_features["hype_bonus"] = df_features.apply(
    lambda row: get_hype(row["team1"], row["team2"]), axis=1
)

df_features["entertainment_score"] = df_features["entertainment_score"] + df_features["hype_bonus"]
```

```python
p85 = df_features["entertainment_score"].quantile(0.85)
p55 = df_features["entertainment_score"].quantile(0.55)
p25 = df_features["entertainment_score"].quantile(0.25)

def categorize(score):
    if score >= p85:
        return "Must Watch"
    elif score >= p55:
        return "Worth Watching"
    elif score >= p25:
        return "Optional"
    else:
        return "Skip"

df_features["category"] = df_features["entertainment_score"].apply(categorize)
print(df_features["category"].value_counts())
print(f"\np85 threshold: {round(p85, 3)}")
print(f"p55 threshold: {round(p55, 3)}")
print(f"p25 threshold: {round(p25, 3)}")
```
